```python
import pandas as pd

df = pd.read_csv("adult11.csv")
df.info()
df.describe(include="all")
df.isnull().sum()
df.nunique()
```

```python
df = df.loc[:, df.nunique() > 1]
df = df.dropna()
df = df.map(lambda x : x.strip().lower() if isinstance(x, str) else x)
for col in df.select_dtypes(include="object").columns:
  df[col] = df[col].astype("category")
```

```python
df['salary'].value_counts()
df['gender'].value_counts(normalize=True)
df[df['salary'] == '>50k']['age'].mean()
df['occupation'].value_counts().head(10)
df.groupby('education')['salary'].value_counts(normalize=True)
df.groupby('salary')['hours-per-week'].mean()
df[df['salary'] == '>50k']['native-country'].value_counts(normalize=True).head(10)
df.groupby('salary')['age'].describe()
df.groupby('gender')['salary'].value_counts(normalize=True)
df[['age', 'hours-per-week']].corr()
```
